```python

import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat.webp")
cv2_imshow(image)
```

```python
print("shape", image.shape)
# height, width, color channels -> (BGR) blue, green, red

print("height:", image.shape[0])
print("width:", image.shape[1])
print("channels:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("blue", b, "green", g, "red", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray, (500, 300))
cv2_imshow(resized)
```

```python
cropped = gray[50:100, 50:100]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite("flipped.png", flipped)
```

```python
cv2.imwrite("cropped_cat.png", cropped)
```
