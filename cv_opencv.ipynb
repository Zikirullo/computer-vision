```python

import cv2
```

```python
from google.colab.patches import cv2_imshow

image = cv2.imread("/content/cat.webp")
cv2_imshow(image)
```

```python
print("shape", image.shape)
# height, width, color channels -> (BGR) blue, green, red

print("height:", image.shape[0])
print("width:", image.shape[1])
print("channels:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("blue", b, "green", g, "red", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
cv2_imshow(gray)
```

```python

resized = cv2.resize(gray,(300,300))
cv2_imshow(resized)
```

```python

cropped = gray[90:300, 100:350]
cv2_imshow(cropped)
```

```python

rotated = cv2.rotate(gray, cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(rotated)
```

```python

flipped = cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python

cv2.imwrite("flipped_cat.jpg", flipped)
```

**`Open CV Video Streaming Codes`**

```python
import cv2
print("Hello World!")

cap = cv2.VideoCapture(0)

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

while True:
    success, frame = cap.read()

    if not success:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.5, 4)

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv2.imshow("Real time face detection", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()

# cap = cv2.VideoCapture(0)

# while True:
#     success, frame = cap.read()

#     if not success:
#         break

#     # gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
#     # blurred = cv2.GaussianBlur(frame, (15, 15), 2)
#     # edges1 = cv2.Canny(frame, 100, 200)
#     # edges2 = cv2.Canny(frame, 50, 250)

#     flipped = cv2.flip(frame, 1)

#     # cv2.imshow("original", frame)
#     # cv2.imshow("gray", gray)
#     # cv2.imshow("blurred", blurred)
#     # cv2.imshow("edges1", edges1)
#     # cv2.imshow("edges2", edges2)
#     # cv2.imshow("flipped", flipped)

#     if cv2.waitKey(1) & 0xFF == ord("q"):
#         break

# cap.release()
# cv2.destroyAllWindows()

# ========================\
```
