```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line Plot

```python
x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7, 8])

plt.plot(x, y)
plt.title("Basic line plot")
plt.xlabel("X-Values")
plt.ylabel("Y-Values")
plt.show()
```

# Styled Line

```python

x = np.array([1, 2, 3, 4])
y = np.array([11, 5, 33, 44])

plt.plot(y, linestyle="dotted", color="red", marker="o")
plt.title("Styled Line Example")
plt.show()
```

# Multple Lines

```python
x = np.array([2, 5, 19, 8])
y1 = np.array([22, 51, 1, 4])
y2 = np.array([7, 1, 39, 6])

plt.plot(x, y1, label="line 1", color="red", marker="s")
plt.plot(x, y2, color="blue", label="line 2", marker="o", linestyle="--")
plt.title("Multple Lines Example")
plt.xlabel("X-value")
plt.ylabel("Y-value")
plt.legend()
plt.show()
```

# **Barchart with Numpy and Matplotlib**

```python
x = np.array(["A", "B", "C", "D", "E", "F", "G"])
y = np.array([1, 2, 5, 4, 1, 2, 1])

plt.bar(x, y, color="orange")
plt.title("Bar Chart example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

# **Scatter plot with numpy and matplotlib**

```python
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([7, 8, 9, 1, 2, 3])

plt.scatter(x, y, color="purple", s=200)
plt.title("Scatter Plot example")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

# Histogram with Numpu and Matplotlib

```python
data = np.array([1, 2, 3, 4, 5, 6, 7])
plt.hist(data, bins=5, color="skyblue", edgecolor="black")
plt.xlabel("Values")
plt.ylabel("Counts")
plt.show()
```

# Subplots(Multple Graphs Together)

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])

plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.plot(x, y1, color="red")
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.bar(x, y2, color="blue")
plt.title("Plot 2")

plt.tight_layout()
plt.show()
plt.savefig('my_subplots.png')
```
